# 05 · Insights export

Энэ notebook нь `src/insights.py`-г дуудаж scrollytelling website-ийн хэрэглэх
`website/data/insights.json`-ыг үүсгэнэ.


```python
import subprocess
import sys
from pathlib import Path

result = subprocess.run(
    [sys.executable, str(Path('..') / 'src' / 'insights.py')],
    capture_output=True, text=True, encoding='utf-8',
)
print(result.stdout)
if result.returncode != 0:
    print('ALDAA:', result.stderr)
```

```python
# JSON-г уншаад товчлол харуулах
import json
from pathlib import Path

p = Path('..') / 'website' / 'data' / 'insights.json'
with open(p, encoding='utf-8') as f:
    data = json.load(f)
print(f'Хэмжээ: {p.stat().st_size // 1024} KB')
print(f'Гол түлхүүрүүд: {list(data.keys())}')
```

```python
# Story moments — website-ийн гол тоонууд
print(json.dumps(data['story_moments'], ensure_ascii=False, indent=2))
```
